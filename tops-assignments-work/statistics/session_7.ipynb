#### 1. Given the following data from an e-commerce app: out of 500 users, 120 bought phone accessories, 80 bought a phone, and 50 bought both, calculate the probability that a user bought a phone given they already bought accessories (P(Phone|Accessories)).

```python
import numpy as np
import pandas as pd

data = pd.DataFrame({
    "Total Users": [500],
    "Accessories Bought": [120],
    "Phone Bought": [80],
    "Bought Both": [50]
})

print(data)

# P(Phone | Accessories)
prob_phone_given_accessories = (
    data.loc[0, "Bought Both"] /
    data.loc[0, "Accessories Bought"]
)

print("\nFormula:")
print("P(Phone | Accessories) = P(Both) / P(Accessories)")

print("Calculation:")
print(f"= {data.loc[0,'Bought Both']} / {data.loc[0,'Accessories Bought']}")

print(f"\nAnswer = {prob_phone_given_accessories:.4f}")
print(f"= {prob_phone_given_accessories*100:.2f}%")
```

#### 3. Write a Python function bayes_churn_probability(prior_churn, prob_email_given_churn, prob_email) that uses Bayes’ Theorem to compute the probability a user will churn given they received a 'we miss you' email. Test your function with prior_churn=0.15, prob_email_given_churn=0.6, prob_email=0.25.

```python
def bayes_churn_probability(prior_churn,
                            prob_email_given_churn,
                            prob_email):

    return (prob_email_given_churn * prior_churn) / prob_email


prior_churn = 0.15
prob_email_given_churn = 0.60
prob_email = 0.25

posterior = bayes_churn_probability(
    prior_churn,
    prob_email_given_churn,
    prob_email
)

df = pd.DataFrame({
    "Prior Churn": [prior_churn],
    "P(Email | Churn)": [prob_email_given_churn],
    "P(Email)": [prob_email],
    "P(Churn | Email)": [posterior]
})

print(df)

print(f"\nAnswer = {posterior:.2f}")
print(f"= {posterior*100:.2f}%")
```

#### 4. A digital wallet app detects fraud based on spending patterns. If 2% of all transactions are fraudulent, and the app’s alert system correctly flags 90% of fraudulent transactions but also falsely flags 5% of normal transactions, use Bayes’ Theorem to calculate the probability that a transaction is actually fraudulent when flagged by the alert system.<br><br><em><strong>Hint:</strong> Define events: F = fraud, A = alert. Use Bayes’ formula: P(F|A) = [P(A|F) * P(F)] / P(A).</em>

```python
P_F = 0.02
P_A_given_F = 0.90
P_A_given_notF = 0.05

P_notF = 1 - P_F

P_A = (P_A_given_F * P_F) + (P_A_given_notF * P_notF)

P_F_given_A = (P_A_given_F * P_F) / P_A

fraud = pd.DataFrame({
    "Fraud Probability": [P_F],
    "Alert Given Fraud": [P_A_given_F],
    "False Alert": [P_A_given_notF],
    "P(Fraud | Alert)": [P_F_given_A]
})

print(fraud)

print("\nFormula:")
print("P(Fraud|Alert) = (P(Alert|Fraud) × P(Fraud)) / P(Alert)")

print(f"\nAnswer = {P_F_given_A:.4f}")
print(f"= {P_F_given_A*100:.2f}%")
```
