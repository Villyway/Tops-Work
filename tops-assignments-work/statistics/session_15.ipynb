#### 1. Create a CSV file named purchase_data.csv with columns: customer_id, age_group (18-25, 26-35, 36-50), and purchase_value. Enter at least 15 sample records with 5 customers in each age group.

```python
import pandas as pd

data = {
    "customer_id": range(1, 16),
    "age_group": [
        "18-25","18-25","18-25","18-25","18-25",
        "26-35","26-35","26-35","26-35","26-35",
        "36-50","36-50","36-50","36-50","36-50"
    ],
    "purchase_value": [
        1200,1350,1500,1250,1400,
        1800,1900,1750,2000,1850,
        2400,2500,2300,2450,2600
    ]
}

df = pd.DataFrame(data)

df.to_csv("purchase_data.csv", index=False)

print(df)
print("\nCSV file 'purchase_data.csv' created successfully.")
```

#### 2. Write a Python script using pandas and scipy.stats that reads purchase_data.csv and performs a one-way ANOVA to compare the average purchase_value across the three age groups. Print the F-statistic and p-value.

```python
import pandas as pd
from scipy.stats import f_oneway

df = pd.read_csv("purchase_data.csv")

group1 = df[df["age_group"]=="18-25"]["purchase_value"]
group2 = df[df["age_group"]=="26-35"]["purchase_value"]
group3 = df[df["age_group"]=="36-50"]["purchase_value"]

f_stat, p_value = f_oneway(group1, group2, group3)

print("F-Statistic:", round(f_stat,4))
print("P-Value:", round(p_value,4))
```

#### 3. Based on your ANOVA results, interpret whether there is a statistically significant difference in average purchase_value between the age groups. Write your answer in 2-3 lines.

```python
alpha = 0.05

if p_value < alpha:
    print("There is a statistically significant difference")
    print("in average purchase values among the three age groups.")
else:
    print("There is no statistically significant difference")
    print("in average purchase values among the three age groups.")
```

#### 4. Calculate and print the between-group variance and within-group variance for the purchase_value data using numpy or pandas. Explain briefly what each value represents in this context.

**Hint:** Between-group variance measures differences between age group means; within-group variance measures variation inside each age group.

```python
import numpy as np

overall_mean = df["purchase_value"].mean()

group_means = df.groupby("age_group")["purchase_value"].mean()

between_variance = np.var(group_means, ddof=1)

within_variance = df.groupby("age_group")["purchase_value"].var().mean()

print("Between Group Variance:", round(between_variance,2))
print("Within Group Variance:", round(within_variance,2))

print("\nExplanation:")
print("Between-group variance shows differences between age group means.")
print("Within-group variance shows variation within each age group.")
```

#### 5. Imagine you are analyzing Zomato order data for three different customer segments: Students, Working Professionals, and Retirees. Use ChatGPT or Copilot to generate Python code that performs a one-way ANOVA on their average monthly order amounts. Paste the generated code and run it with dummy data for each group.

```python
from scipy.stats import f_oneway

students = [1200,1350,1400,1500,1450]
working = [2200,2100,2300,2250,2400]
retirees = [1700,1800,1750,1650,1850]

f_stat, p_value = f_oneway(students, working, retirees)

print("F-Statistic:", round(f_stat,4))
print("P-Value:", round(p_value,4))

if p_value < 0.05:
    print("The average monthly order amount differs significantly among the three customer segments.")
else:
    print("There is no significant difference among the three customer segments.")
```
