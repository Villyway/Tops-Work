#### 1. Create a contingency table in Excel or Google Sheets showing the relationship between users' favorite food delivery app (Zomato, Swiggy, Uber Eats) and their preferred payment method (Paytm, PhonePe, Credit Card) using a sample of at least 20 responses.

```python
import pandas as pd

# Sample data
data = {
    "App": [
        "Zomato","Swiggy","Uber Eats","Zomato","Swiggy",
        "Uber Eats","Zomato","Swiggy","Uber Eats","Zomato",
        "Swiggy","Uber Eats","Zomato","Swiggy","Uber Eats",
        "Zomato","Swiggy","Uber Eats","Zomato","Swiggy"
    ],
    "Payment": [
        "Paytm","PhonePe","Credit Card","PhonePe","Paytm",
        "Credit Card","Paytm","PhonePe","Paytm","Credit Card",
        "Paytm","PhonePe","Credit Card","Credit Card","PhonePe",
        "Paytm","Paytm","Credit Card","PhonePe","PhonePe"
    ]
}

df = pd.DataFrame(data)

contingency_table = pd.crosstab(df["App"], df["Payment"])

print("Contingency Table")
print(contingency_table)
```

#### 2. Given the following observed counts of music streaming preference among males and females: Spotify (Males: 12, Females: 18), Gaana (Males: 8, Females: 7), YouTube Music (Males: 10, Females: 15), calculate the expected counts for each cell assuming independence.

**Hint:** Use the formula: (row total × column total) / grand total.

```python
import pandas as pd

observed = pd.DataFrame({
    "Male":[12,8,10],
    "Female":[18,7,15]
}, index=["Spotify","Gaana","YouTube Music"])

print("Observed Counts")
print(observed)

row_total = observed.sum(axis=1)
col_total = observed.sum(axis=0)
grand_total = observed.values.sum()

expected = pd.DataFrame(index=observed.index, columns=observed.columns)

for row in observed.index:
    for col in observed.columns:
        expected.loc[row, col] = (row_total[row] * col_total[col]) / grand_total

print("\nExpected Counts")
print(expected)
```

#### 3. Perform a chi-square test of independence using the data from your contingency table in task 1 to determine if there is a significant association between app choice and payment method. Show your calculation steps and state your conclusion.

```python
import pandas as pd
from scipy.stats import chi2_contingency

data = {
    "App": [
        "Zomato","Swiggy","Uber Eats","Zomato","Swiggy",
        "Uber Eats","Zomato","Swiggy","Uber Eats","Zomato",
        "Swiggy","Uber Eats","Zomato","Swiggy","Uber Eats",
        "Zomato","Swiggy","Uber Eats","Zomato","Swiggy"
    ],
    "Payment": [
        "Paytm","PhonePe","Credit Card","PhonePe","Paytm",
        "Credit Card","Paytm","PhonePe","Paytm","Credit Card",
        "Paytm","PhonePe","Credit Card","Credit Card","PhonePe",
        "Paytm","Paytm","Credit Card","PhonePe","PhonePe"
    ]
}

df = pd.DataFrame(data)

table = pd.crosstab(df["App"], df["Payment"])

chi2, p, dof, expected = chi2_contingency(table)

print("Chi-Square Statistic:", round(chi2,4))
print("P-value:", round(p,4))
print("Degrees of Freedom:", dof)

if p < 0.05:
    print("\nReject the Null Hypothesis")
    print("There is a significant association between app choice and payment method.")
else:
    print("\nFail to Reject the Null Hypothesis")
    print("There is no significant association between app choice and payment method.")
```

#### 4. For a new Flipkart product launch, suppose you surveyed 30 users and observed the following: 18 prefer the new product, 12 do not. If the company claims that 70% of users will prefer the new product, perform a chi-square goodness of fit test to check this claim.

**Hint:** Calculate expected counts based on the claimed proportion.

```python
from scipy.stats import chisquare

# Observed values
observed = [18, 12]

# Expected values (70% and 30%)
expected = [21, 9]

chi2, p = chisquare(f_obs=observed, f_exp=expected)

print("Chi-Square Statistic:", round(chi2,4))
print("P-value:", round(p,4))

if p < 0.05:
    print("\nReject the Null Hypothesis")
    print("The observed data does not support the company's claim.")
else:
    print("\nFail to Reject the Null Hypothesis")
    print("The observed data supports the company's claim.")
```
