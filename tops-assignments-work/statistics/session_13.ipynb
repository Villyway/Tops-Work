#### 1. Using Python and the scipy.stats library, perform a one-sample t-test to check if the average delivery time for Zomato orders in your city is significantly different from 30 minutes. Assume you have this sample data: [28, 32, 27, 35, 29, 31, 30, 33, 26, 34].

```python
from scipy.stats import ttest_1samp

delivery_time = [28, 32, 27, 35, 29, 31, 30, 33, 26, 34]

t_stat, p_value = ttest_1samp(delivery_time, 30)

print("T-Statistic:", round(t_stat, 4))
print("P-Value:", round(p_value, 4))

if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

#### 2. Collect or simulate two independent samples of daily step counts: one group for users before installing a fitness app (e.g., Google Fit) and one group after installation. Use Python to perform an independent (two-sample) t-test and determine if the average steps per day increased after using the app.

**Hint:** Use scipy.stats.ttest_ind().

```python
from scipy.stats import ttest_ind

before = [5000, 5200, 4800, 5100, 5300, 4900, 5050, 5150, 4950, 5000]
after = [6200, 6400, 6100, 6500, 6300, 6150, 6250, 6450, 6350, 6200]

t_stat, p_value = ttest_ind(before, after)

print("T-Statistic:", round(t_stat, 4))
print("P-Value:", round(p_value, 4))

if p_value < 0.05:
    print("Average steps significantly increased.")
else:
    print("No significant increase in average steps.")
```

#### 3. Simulate paired data for 10 users showing their average screen time (in hours) before and after enabling 'Focus Mode' on their phones. Use Python to conduct a paired t-test and report whether 'Focus Mode' significantly reduced screen time.

**Hint:** Use scipy.stats.ttest_rel().

```python
from scipy.stats import ttest_rel

before = [6.5, 7.2, 5.8, 6.9, 7.5, 6.8, 7.1, 6.3, 7.0, 6.7]
after = [5.8, 6.3, 5.2, 6.0, 6.7, 6.1, 6.2, 5.8, 6.3, 6.0]

t_stat, p_value = ttest_rel(before, after)

print("T-Statistic:", round(t_stat, 4))
print("P-Value:", round(p_value, 4))

if p_value < 0.05:
    print("Focus Mode significantly reduced screen time.")
else:
    print("Focus Mode did not significantly reduce screen time.")
```

#### 4. Given this scenario: You want to compare the average ratings of two food delivery apps (Zomato vs Swiggy) based on 15 randomly selected user ratings for each app. Write Python code to perform the appropriate t-test, and explain why you chose that test.

```python
from scipy.stats import ttest_ind

zomato = [4.2,4.5,4.0,4.3,4.6,4.4,4.1,4.5,4.2,4.3,4.4,4.6,4.5,4.3,4.2]
swiggy = [4.0,4.1,3.9,4.2,4.1,4.0,4.2,4.1,4.0,4.3,4.1,4.2,4.0,4.1,4.2]

t_stat, p_value = ttest_ind(zomato, swiggy)

print("T-Statistic:", round(t_stat,4))
print("P-Value:", round(p_value,4))

print("\nReason:")
print("Independent t-test is used because ratings are collected from two different groups of users.")
```
