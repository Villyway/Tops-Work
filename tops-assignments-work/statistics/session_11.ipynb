#### 1. Simulate the Central Limit Theorem in Python by taking 1000 random samples (size 30) from a skewed dataset (e.g., number of likes on Instagram posts), calculating the mean of each sample, and plotting the histogram of these sample means using matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Skewed dataset (Exponential Distribution)
likes = np.random.exponential(scale=100, size=10000)

sample_means = []

# Take 1000 samples of size 30
for i in range(1000):
    sample = np.random.choice(likes, size=30)
    sample_means.append(np.mean(sample))

# Plot histogram of sample means
plt.figure(figsize=(8,5))
plt.hist(sample_means, bins=30, edgecolor='black')
plt.title("Central Limit Theorem Simulation")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

#### 4. Modify the CLT simulation code to use a much smaller sample size (n=5) and compare the histogram of sample means to the one with n=30. Briefly describe what changes you observe in the shape of the distribution.

**Hint:** Run both simulations and compare the spread and normality of the histograms.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Skewed dataset
data = np.random.exponential(scale=100, size=10000)

sample_means_5 = []
sample_means_30 = []

# Sample size = 5
for i in range(1000):
    sample = np.random.choice(data, size=5)
    sample_means_5.append(np.mean(sample))

# Sample size = 30
for i in range(1000):
    sample = np.random.choice(data, size=30)
    sample_means_30.append(np.mean(sample))

plt.figure(figsize=(8,5))
plt.hist(sample_means_5, bins=30, alpha=0.6, label="n = 5")
plt.hist(sample_means_30, bins=30, alpha=0.6, label="n = 30")

plt.title("CLT Comparison")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.legend()
plt.grid(True)

plt.show()

print("""
Observation:

1. The histogram for n=5 is wider and more skewed.
2. The histogram for n=30 is more bell-shaped.
3. Larger sample sizes produce a distribution closer to normal.
""")
```

#### 5. Use ChatGPT to generate Python code that demonstrates the Central Limit Theorem using random samples from a non-normal distribution (e.g., exponential). Run the code, paste the output plot, and briefly describe what you observe.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Non-normal (Exponential) distribution
population = np.random.exponential(scale=50, size=10000)

sample_means = []

# Generate 1000 sample means
for i in range(1000):
    sample = np.random.choice(population, size=30)
    sample_means.append(np.mean(sample))

plt.figure(figsize=(8,5))
plt.hist(sample_means, bins=30, edgecolor="black")

plt.title("Central Limit Theorem using Exponential Distribution")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.grid(True)

plt.show()

print("""
Observation:

Although the original exponential distribution is highly skewed,
the distribution of sample means becomes approximately normal
when the sample size is 30. This demonstrates the Central Limit Theorem.
""")
```
