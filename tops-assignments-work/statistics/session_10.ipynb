#### 2. Given an array of 1000 Flipkart orders labeled by city (e.g., 'Ahmedabad', 'Surat', 'Vadodara'), write code to generate a stratified sample of 60 orders, ensuring each city is proportionally represented in the sample.

**Hint:** Calculate the proportion of orders per city and sample accordingly from each group.

```python
import numpy as np
import pandas as pd

np.random.seed(42)

# Create 1000 orders with city labels
cities = np.random.choice(
    ["Ahmedabad", "Surat", "Vadodara"],
    size=1000,
    p=[0.5, 0.3, 0.2]
)

orders = pd.DataFrame({
    "Order ID": range(1, 1001),
    "City": cities
})

print("Original Data:")
print(orders["City"].value_counts())

# Stratified Sample
sample_size = 60

stratified_sample = (
    orders.groupby("City", group_keys=False)
    .apply(lambda x: x.sample(round(len(x) / len(orders) * sample_size), random_state=42))
)

print("\nStratified Sample:")
print(stratified_sample)

print("\nSample Count:")
print(stratified_sample["City"].value_counts())
```

#### 3. Write a function to perform systematic sampling on a list of 500 Zomato restaurant reviews, selecting every 10th review to create a sample set.

```python
import pandas as pd

# Create 500 dummy reviews
reviews = [f"Review {i}" for i in range(1, 501)]

# Function for Systematic Sampling
def systematic_sampling(data, step):
    return data[::step]

# Select every 10th review
sample_reviews = systematic_sampling(reviews, 10)

df = pd.DataFrame({
    "Sample Reviews": sample_reviews
})

print(df)
print("\nTotal Sampled Reviews:", len(df))
```

#### 4. Suppose you want to estimate the average delivery time for Swiggy orders in your city. Explain with calculations how the sample size affects the accuracy of your estimate if you have access to 10,000 delivery records and can only sample 50, 200, or 500 records.

```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Generate 10,000 delivery times (in minutes)
delivery_times = np.random.normal(loc=30, scale=5, size=10000)

df = pd.DataFrame({
    "Delivery Time": delivery_times
})

# Sample sizes
sample_50 = df.sample(n=50, random_state=42)
sample_200 = df.sample(n=200, random_state=42)
sample_500 = df.sample(n=500, random_state=42)

print("Population Mean :", round(df["Delivery Time"].mean(), 2))
print("Sample Mean (50) :", round(sample_50["Delivery Time"].mean(), 2))
print("Sample Mean (200):", round(sample_200["Delivery Time"].mean(), 2))
print("Sample Mean (500):", round(sample_500["Delivery Time"].mean(), 2))

print("\nConclusion:")
print("Larger sample sizes generally provide estimates closer to the population mean.")
```

#### 5. Modify the stratified sampling demo shown in class: instead of customer segments, use a dataset of IPL cricket fans grouped by favorite team, and generate a stratified sample of 40 fans for a survey, ensuring each team is represented proportionally.

**Constraint:** Do not copy the code from the demo—write your own logic for grouping and sampling.

```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Create dataset of 200 IPL fans
teams = np.random.choice(
    ["CSK", "MI", "RCB", "KKR", "GT"],
    size=200,
    p=[0.25, 0.25, 0.20, 0.15, 0.15]
)

fans = pd.DataFrame({
    "Fan ID": range(1, 201),
    "Favorite Team": teams
})

print("Original Team Distribution:")
print(fans["Favorite Team"].value_counts())

# Stratified sample of 40 fans
sample_size = 40

stratified_sample = (
    fans.groupby("Favorite Team", group_keys=False)
        .apply(lambda x: x.sample(
            n=max(1, round(len(x) / len(fans) * sample_size)),
            random_state=42
        ))
)

print("\nStratified Sample:")
print(stratified_sample)

print("\nSample Distribution:")
print(stratified_sample["Favorite Team"].value_counts())
```
