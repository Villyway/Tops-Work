#### 4. Write a function or pseudocode that takes a column of data (array of values) from a Flipkart order history and automatically detects whether it is likely Nominal, Ordinal, Interval, or Ratio scale. Briefly describe the logic used for each type.

**Constraint:** You do not need to write a perfect algorithm — focus on simple rules based on value types and patterns.

```python
def detect_scale(data):

    if all(isinstance(x, str) for x in data):
        print("Likely Nominal Scale")

    elif all(isinstance(x, (int, float)) for x in data):

        if min(data) >= 0:
            print("Likely Ratio Scale")
        else:
            print("Likely Interval Scale")

    else:
        print("Scale could not be determined")


payment = ["UPI", "Cash", "Card", "UPI"]
price = [500, 1200, 750, 900]
temperature = [-5, 10, 20, 35]

print("Payment Method:")
detect_scale(payment)

print("\nProduct Price:")
detect_scale(price)

print("\nTemperature:")
detect_scale(temperature)
```
