#### 1. Create an array of 7 daily step counts (e.g., from your phone's fitness app) and write code to calculate the mean (average) number of steps.

```python
import numpy as np

steps = np.array([7200, 8100, 6500, 9000, 7800, 8300, 7600])

mean_steps = np.mean(steps)

print("Daily Steps:", steps)
print("Mean (Average) Steps:", mean_steps)
```

#### 2. Given this list of food delivery times in minutes for your last 9 Zomato/Swiggy orders: [32, 28, 29, 45, 30, 31, 60, 30, 29], write a function to find the median delivery time.

```python
import numpy as np

delivery_times = [32, 28, 29, 45, 30, 31, 60, 30, 29]

def find_median(data):
    return np.median(data)

median = find_median(delivery_times)

print("Delivery Times:", delivery_times)
print("Median Delivery Time:", median)
```

#### 3. Imagine you tracked the genres of 12 YouTube videos you watched this week: ['Music', 'Vlog', 'Music', 'Tech', 'Music', 'Vlog', 'Tech', 'Music', 'Comedy', 'Vlog', 'Music', 'Comedy']. Write code to determine the mode (most watched genre).

```python
from statistics import mode

genres = [
    'Music', 'Vlog', 'Music', 'Tech',
    'Music', 'Vlog', 'Tech', 'Music',
    'Comedy', 'Vlog', 'Music', 'Comedy'
]

most_watched = mode(genres)

print("Genres:", genres)
print("Mode (Most Watched Genre):", most_watched)
```

#### 4. Take this array of daily UPI transaction amounts (in ₹): [100, 120, 105, 110, 5000, 115, 108]. Calculate both the mean and the median, then explain in 2-3 lines which is a better measure of your 'typical' transaction and why.

**Hint:** Consider the effect of the ₹5000 value.

```python
import numpy as np

transactions = np.array([100, 120, 105, 110, 5000, 115, 108])

mean_value = np.mean(transactions)
median_value = np.median(transactions)

print("Transactions:", transactions)
print("Mean:", mean_value)
print("Median:", median_value)

print("\nExplanation:")
print("The value ₹5000 is an outlier that increases the mean significantly.")
print("The median is less affected by outliers, so it better represents the typical transaction amount.")
```
