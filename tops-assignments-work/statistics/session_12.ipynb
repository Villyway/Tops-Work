#### 1. State the null and alternative hypotheses for testing whether a new cashback offer on a food delivery app (like Zomato) has increased the average number of daily orders compared to before the offer.

```python
print("Hypothesis Testing")

print("\nNull Hypothesis (H0):")
print("The new cashback offer has NOT increased the average number of daily orders.")

print("\nAlternative Hypothesis (H1):")
print("The new cashback offer HAS increased the average number of daily orders.")
```

#### 2. Given the following scenario: A music streaming app claims their new playlist recommendation algorithm increases average listening time from 40 to 45 minutes. If you set up a hypothesis test, identify which type of error (Type I or Type II) would occur if you wrongly conclude the algorithm works when it actually doesn't, and explain in one sentence why.

Ans:- Type I Error
Reason:
A Type I Error occurs when we reject the Null Hypothesis even though
it is actually true.

Here, we wrongly conclude that the new recommendation algorithm
works when in reality it does not.

#### 3. Suppose you run an A/B test on a shopping app (like Flipkart) and get a p-value of 0.03 when testing if a new discount banner increases purchases. At a 95% confidence level, state whether you reject or fail to reject the null hypothesis and explain what this means in the context of the app.

```python
p_value = 0.03
alpha = 0.05

print("P-value:", p_value)
print("Significance Level:", alpha)

if p_value < alpha:
    print("\nDecision: Reject the Null Hypothesis")
    print("Conclusion: The new discount banner significantly increases purchases.")
else:
    print("\nDecision: Fail to Reject the Null Hypothesis")
    print("Conclusion: There is not enough evidence that the banner increases purchases.")
```

#### 4. For a cricket app's push notification feature, you want to be 99% confident that sending match alerts increases user engagement. Explain what the 99% confidence level means in this context, and how it affects your decision to accept or reject the feature.

A 99% confidence level means we are highly confident that
the result is not due to random chance.
