#### 1. Open Microsoft Excel or Google Sheets, create a column named 'Order Amount', and enter 20 random numbers between 100 and 1000 to simulate Zomato order values for a day.

```python
import pandas as pd
import numpy as np

np.random.seed(42)

order_amount = np.random.randint(100, 1001, 20)

df = pd.DataFrame({
    "Order Amount": order_amount
})

print(df)


df.to_csv("order_amount.csv", index=False)
```

#### 2. From your 'Order Amount' data, calculate the mean, median, and mode using Excel or Sheets functions (AVERAGE, MEDIAN, MODE) and write the results below your data.

```python
import pandas as pd

df = pd.read_csv("order_amount.csv")

print(df)

print("\nMean:", df["Order Amount"].mean())
print("Median:", df["Order Amount"].median())
print("Mode:", df["Order Amount"].mode()[0])
```

#### 3. Select the first 5 values from your 'Order Amount' column as a sample. Calculate the sample mean and compare it to the population mean you found earlier.

**Hint:** Use the AVERAGE function for both calculations and observe the difference between sample and population means.

```python
sample = df["Order Amount"].head(5)

sample_mean = sample.mean()
population_mean = df["Order Amount"].mean()

print("Sample Values:")
print(sample)

print("\nSample Mean:", sample_mean)
print("Population Mean:", population_mean)

print("\nDifference:", abs(sample_mean - population_mean))
```

#### 4. Write a short paragraph explaining the difference between 'population' and 'sample' using your Zomato order data as an example. Mention which one is more practical to use for a food delivery company and why.

Ans:-
Population:
The population includes all 20 Zomato order amounts collected for the day.

Sample:
A sample is a smaller subset of the population, such as the first 5 order amounts.

In real-world food delivery companies, using a sample is often more practical
because collecting and analyzing the entire population can be time-consuming
and expensive. A well-chosen sample can provide useful insights about the
