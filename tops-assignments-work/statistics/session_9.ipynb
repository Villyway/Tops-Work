#### 1. Write a Python script that generates 1000 random numbers following a normal distribution with mean 500 and standard deviation 100, then plot the resulting distribution using matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt

mean = 500
std_dev = 100

sales = np.random.normal(mean, std_dev, 1000)

plt.figure(figsize=(8,5))
plt.hist(sales, bins=30)
plt.title("Normal Distribution of Sales Amount")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")
plt.grid(True)

plt.show()
```

#### 2. Given an array of daily step counts from a fitness app, calculate the Z-score for a specific day where the user walked 9500 steps. Assume the mean is 8000 and standard deviation is 1200.<br><br><em><strong>Hint:</strong> Use the formula Z = (X - μ) / σ.</em>

```python
mean = 8000
std_dev = 1200
steps = 9500

z_score = (steps - mean) / std_dev

print("Mean =", mean)
print("Standard Deviation =", std_dev)
print("Steps =", steps)
print("Z-score =", round(z_score,2))
```

#### 3. Plot a standard normal curve (mean 0, standard deviation 1) using matplotlib, and shade the area between Z = -1 and Z = 1 to visualize the probability of values within one standard deviation.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(-4, 4, 1000)
y = norm.pdf(x, 0, 1)

plt.figure(figsize=(8,5))
plt.plot(x, y)

plt.fill_between(
    x,
    y,
    where=(x >= -1) & (x <= 1),
    alpha=0.4
)

plt.title("Standard Normal Distribution")
plt.xlabel("Z-score")
plt.ylabel("Probability Density")
plt.grid(True)

plt.show()
```

#### 4. You are analyzing the delivery times (in minutes) for a food delivery app like Zomato. Given a normal distribution with mean 30 and standard deviation 5, write code to find the probability that a randomly selected delivery takes less than 25 minutes.<br><br><em><strong>Hint:</strong> Use scipy.stats.norm.cdf().</em>

```python
from scipy.stats import norm

mean = 30
std_dev = 5

probability = norm.cdf(25, mean, std_dev)

print("Mean =", mean)
print("Standard Deviation =", std_dev)
print("Probability =", probability)
print("Percentage =", round(probability*100,2),"%")
```
