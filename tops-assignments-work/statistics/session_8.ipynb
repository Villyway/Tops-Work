#### 1. Simulate a Bernoulli trial in Python to model whether a user successfully logs into their Zomato account (success=1, failure=0) with a given probability of success p=0.85. Run the simulation 20 times and print the results.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson

# Task 1

p = 0.85

results = np.random.binomial(1, p, 20)

df = pd.DataFrame({
    "Attempt": range(1, 21),
    "Login Result": results
})

print(df)
```

####2. Write a Python function that uses the binomial distribution to calculate the probability of getting exactly 3 successful UPI payments out of 5 attempts on Paytm, if the success rate per payment is 0.9.<br><br><em><strong>Hint:</strong> Use scipy.stats.binom.pmf for the calculation.</em>

```python
# Task 2

n = 5
k = 3
p = 0.9

probability = binom.pmf(k, n, p)

print("Probability =", probability)
```

#### 3. Suppose the average number of new playlist creations per minute on Spotify is 4. Use the Poisson distribution to calculate the probability that exactly 6 playlists are created in a given minute.<br><br><em><strong>Hint:</strong> Use scipy.stats.poisson.pmf.</em>

```python
# Task 3

lam = 4
k = 6

probability = poisson.pmf(k, lam)

print("Probability =", probability)
```

#### 4. Given a scenario where Flipkart receives an average of 10 product reviews per hour, write code to simulate and plot the number of reviews received in 12 randomly selected hours using the Poisson distribution.<br><br><em><strong>Hint:</strong> Use numpy.random.poisson and matplotlib for plotting.</em>

```python
# Task 4

np.random.seed(42)

reviews = np.random.poisson(lam=10, size=12)

hours = np.arange(1, 13)

df = pd.DataFrame({
    "Hour": hours,
    "Reviews": reviews
})

print(df)

plt.figure(figsize=(8,4))
plt.plot(hours, reviews, marker="o")
plt.title("Product Reviews per Hour")
plt.xlabel("Hour")
plt.ylabel("Reviews")
plt.grid(True)
plt.show()
```
