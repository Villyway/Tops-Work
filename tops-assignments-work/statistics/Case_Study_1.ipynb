### Step 2: Create a PivotTable to *analyze total sales* (Quantity * UnitPrice) by Country. **Identify the top 3 countries by total sales** and write their names in a cell below your table.

### Step 3: Calculate the average order value (*total sales per InvoiceNo*) for all customers. Highlight any InvoiceNo where the order value is above ₹10,000.

```python
''' Checking the relationship between InvoiceNo and CustomerID 
Question says to calculate avg order value = total sales per InvoiceNo. So i need to check whether 1 InvoiceNo = 1 Customer'''

check = retail_data.groupby('InvoiceNo')['CustomerID'].nunique()

# If this is 1, each invoice belongs to a single customer
if check.max() == 1:
        print('Each invoice belongs to a single customer')  
```

```python
''' Calculate total sales per InvoiceNo '''

invoice_sales = retail_data.groupby('InvoiceNo',as_index=False).agg(Order_Value = ('Sales','sum'))
```

```python
''' Highlight invoices above ₹10,000 '''

high_value_invoices = invoice_sales[invoice_sales['Order_Value'] > 10000]
```

```python
''' Calculate average order value across all invoices or for all customers '''

avg_order_value = invoice_sales['Order_Value'].mean()

print("Average Order Value:", round(avg_order_value,2))
print("\nInvoices above 10,000:\n", high_value_invoices)
```

### Step 4: Perform a t-test to compare the *average order value between customers from 'France' and 'Germany'*. State whether the difference is statistically significant at the 0.05 level

```python
invoice_sales_country_wise = retail_data.groupby(['InvoiceNo','Country'],as_index=False).agg(Order_Value = ('Sales','sum'))
```

```python
france_orders = invoice_sales_country_wise[invoice_sales_country_wise['Country'] == 'France']['Order_Value']
germany_orders = invoice_sales_country_wise[invoice_sales_country_wise['Country'] == 'Germany']['Order_Value']
```

```python
''' France vs Germany side by side with their averages and invoice counts '''

avg_df = pd.DataFrame({
    'Country': ['France', 'Germany'],
    'Average_Order_Value': [france_orders.mean(), germany_orders.mean()],
    'Invoice_Count': [france_orders.count(), germany_orders.count()]
})

print(avg_df)
```

```python
from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(france_orders, germany_orders, equal_var=False)
alpha = 0.05   # Given in the problem statement
print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < alpha:
    print('The difference in average order value between France and Germany is statistically significant')
else:
    print('The difference is not significant at the 0.05 level')
```
