#### 1. Download a small dataset of 10-15 songs with features like danceability, energy, and popularity (you can create a CSV manually or use any open dataset), then calculate the Pearson correlation coefficient between danceability and popularity using pandas.

```python
import pandas as pd

songs = pd.DataFrame({
    "Danceability": [0.80,0.75,0.60,0.90,0.65,0.70,0.85,0.78,0.68,0.88],
    "Energy": [0.90,0.80,0.70,0.95,0.60,0.75,0.85,0.88,0.65,0.92],
    "Popularity": [95,90,70,98,72,80,92,89,75,97]
})

songs.to_csv("songs.csv", index=False)

df = pd.read_csv("songs.csv")

correlation = df["Danceability"].corr(df["Popularity"])

print(df)
print("\nPearson Correlation:", round(correlation,4))
```

#### 2. Given a list of IPL players with their total runs and number of Instagram followers, use scipy.stats.spearmanr to calculate the Spearman correlation between runs and followers, and interpret whether higher runs generally mean more followers.

```python
import pandas as pd
from scipy.stats import spearmanr

players = pd.DataFrame({
    "Player":["A","B","C","D","E","F"],
    "Runs":[5000,7200,6800,4500,8000,6200],
    "Followers":[3.5,6.8,5.9,2.9,8.2,5.4]
})

correlation, p_value = spearmanr(players["Runs"], players["Followers"])

print(players)

print("\nSpearman Correlation:", round(correlation,4))
print("P-value:", round(p_value,4))

if correlation > 0:
    print("Higher runs generally mean more followers.")
else:
    print("No positive relationship found.")
```

#### 3. Create a correlation matrix for three Flipkart product features: price, user rating, and number of reviews, using pandas' corr() method, and display the result.

```python
import pandas as pd

products = pd.DataFrame({
    "Price":[1000,1500,2000,2500,3000,3500,4000],
    "Rating":[4.1,4.3,4.5,4.2,4.6,4.4,4.7],
    "Reviews":[120,180,250,220,310,280,350]
})

corr_matrix = products.corr()

print("Correlation Matrix")
print(corr_matrix)
```

#### 4. Visualize the correlation matrix you created in the previous task as a heatmap using seaborn's heatmap() function, and customize it by adding annotations and a color bar.

**Hint:** Use sns.heatmap(corr_matrix, annot=True, cmap='coolwarm').

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

#### 5. List two potential pitfalls when interpreting correlation coefficients in data from real apps like Zomato or YouTube, and give a short example for each showing how correlation can be misleading.

Pitfall 1:
Correlation does not imply causation.

Example:
Users who spend more on Zomato may also order more frequently,
but spending more does not necessarily cause higher order frequency.

--------------------------------------------------------

Pitfall 2:
A third variable may influence both variables.

Example:
YouTube videos with more views often have more likes,
but both may actually be influenced by better content quality
or stronger promotion.
