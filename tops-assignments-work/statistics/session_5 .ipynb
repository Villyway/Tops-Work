#### 1. Given a list of delivery times (in minutes) for 20 Zomato orders, calculate the 25th, 50th, and 75th percentiles using Python's numpy.percentile function.

```python
import numpy as np

delivery_times = [25, 28, 30, 32, 27, 35, 40, 45, 33, 31,
                  29, 36, 38, 42, 26, 34, 37, 39, 41, 43]

p25 = np.percentile(delivery_times, 25)
p50 = np.percentile(delivery_times, 50)
p75 = np.percentile(delivery_times, 75)

print("25th Percentile (Q1):", p25)
print("50th Percentile (Median):", p50)
print("75th Percentile (Q3):", p75)
```

#### 2. Create a Python script that takes an array of IPL player scores and divides them into quartiles using numpy.quantile. Print the values for Q1, Q2 (median), and Q3.

```python
import numpy as np

scores = [35, 42, 67, 55, 89, 102, 78, 91, 45, 60,
          73, 88, 95, 110, 50]

q1 = np.quantile(scores, 0.25)
q2 = np.quantile(scores, 0.50)
q3 = np.quantile(scores, 0.75)

print("Q1:", q1)
print("Q2 (Median):", q2)
print("Q3:", q3)
```

#### 3. Given a list of daily step counts from your fitness app for one month, calculate the Interquartile Range (IQR) in Python and explain what it tells you about your activity consistency.

**Hint:** IQR = Q3 − Q1

```python
import numpy as np

steps = [6500,7000,7200,6800,7500,8100,8300,7900,7600,7400,
         7700,8200,8500,8000,7800,7300,7100,6900,7600,8100,
         8400,8700,8800,8600,7900,7500,7200,7000,6800,7600]

q1 = np.percentile(steps, 25)
q3 = np.percentile(steps, 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)

print("\nInterpretation:")
print("A smaller IQR indicates more consistent daily activity.")
```

#### 4. Write a Python function to detect outliers in a list of Swiggy order amounts using the IQR method (any value below Q1 - 1.5*IQR or above Q3 + 1.5*IQR). Print the outlier values.

```python
import numpy as np

orders = [250,300,280,320,310,290,305,315,295,285,
          1000,270,325,330,340]

def detect_outliers(data):
    q1 = np.percentile(data, 25)
    q3 = np.percentile(data, 75)

    iqr = q3 - q1

    lower = q1 - (1.5 * iqr)
    upper = q3 + (1.5 * iqr)

    outliers = [x for x in data if x < lower or x > upper]

    return outliers

print("Outliers:", detect_outliers(orders))
```

#### 5. Using matplotlib, create a boxplot of your weekly screen time (in hours) for the past 8 weeks. Label the axes and highlight any outliers.
