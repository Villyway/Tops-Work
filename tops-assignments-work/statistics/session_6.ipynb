#### 1. List all possible outcomes (sample space) for rolling a standard six-sided die and flipping a coin at the same time. Write your answer as a list of ordered pairs.

```python
import numpy as np
import pandas as pd

coin = np.array(['H', 'T'])
dice = np.arange(1, 7)

sample_space = [(c, d) for c in coin for d in dice]

df_sample = pd.DataFrame(sample_space, columns=['Coin', 'Die'])

print("Task 1: Sample Space")
print(df_sample)

print("Total Outcomes:", len(df_sample))
```

#### 2. Suppose you have a deck of 10 cards numbered 1 to 10. Calculate the classical probability of drawing a card with an even number. Show your calculation steps.

```python
cards = pd.DataFrame({'Card': np.arange(1, 11)})

even_cards = cards[cards['Card'] % 2 == 0]

prob_even = len(even_cards) / len(cards)

print("\nTask 2")
print(cards)
print("\nEven Cards")
print(even_cards)
print("Probability =", prob_even)
```

### 3. In a playlist of 8 songs on Spotify, 3 are by Arijit Singh. If you shuffle and play one random song, what is the probability that it is NOT by Arijit Singh? Show your working.

```python
songs = pd.DataFrame({
    'Song': np.arange(1, 9),
    'Artist': ['Arijit', 'Arijit', 'Arijit',
               'Other', 'Other', 'Other',
               'Other', 'Other']
})

not_arijit = songs[songs['Artist'] != 'Arijit']

prob_not_arijit = len(not_arijit) / len(songs)

print("\nTask 3")
print(songs)
print("Probability (Not Arijit) =", prob_not_arijit)
```

### 4. A Zomato user places an order. The probability they order Pizza is 0.3, and the probability they order Burger is 0.4. If the probability they order both is 0.1, use the addition rule to find the probability the user orders Pizza or Burger.<br><br><em><strong>Hint:</strong> P(A or B) = P(A) + P(B) - P(A and B)</em>

```python
P_A = 0.3
P_B = 0.4
P_A_and_B = 0.1

P_A_or_B = P_A + P_B - P_A_and_B

print("\nTask 4")
print("P(Pizza OR Burger) =", P_A_or_B)
```

### 5. A Flipkart user is browsing electronics. The probability they add a phone to cart is 0.2. If you know they are interested in mobiles, the probability rises to 0.5. Briefly explain in your own words what this conditional probability means, using this scenario.

```python
normal_probability = 0.2
conditional_probability = 0.5

print("\nTask 5")
print("Normal Probability :", normal_probability)
print("Conditional Probability :", conditional_probability)

if conditional_probability > normal_probability:
    print("Knowing the user is interested in mobiles increases the probability.")
```
