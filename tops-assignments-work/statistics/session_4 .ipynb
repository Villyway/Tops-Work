#### 1. Given the daily step counts of a user for the last 7 days as [6500, 7200, 8000, 6000, 9000, 7500, 8200], calculate the range of steps and explain what this tells you about the user's activity pattern.

```python
import numpy as np

steps = np.array([6500, 7200, 8000, 6000, 9000, 7500, 8200])

range_steps = np.max(steps) - np.min(steps)

print("Daily Steps:", steps)
print("Maximum Steps:", np.max(steps))
print("Minimum Steps:", np.min(steps))
print("Range:", range_steps)

print("\nExplanation:")
print("The range shows the difference between the highest and lowest")
print("number of steps, indicating how much the user's activity varies.")
```

#### 2. For a list of 5 food delivery order amounts on Zomato: [250, 400, 350, 300, 500], write code to compute the variance and standard deviation of the order amounts.

```python
import numpy as np

orders = np.array([250, 400, 350, 300, 500])

variance = np.var(orders)
std_dev = np.std(orders)

print("Order Amounts:", orders)
print("Variance:", round(variance, 2))
print("Standard Deviation:", round(std_dev, 2))
```

#### 3. You have the monthly mobile data usage (in GB) for a user over 6 months: [3.2, 2.8, 3.5, 3.0, 2.9, 3.1]. Calculate the coefficient of variation (CV) and interpret whether the user's data usage is consistent or highly variable.

**Hint:** CV = (Standard Deviation / Mean) × 100%

```python
import numpy as np

usage = np.array([3.2, 2.8, 3.5, 3.0, 2.9, 3.1])

mean = np.mean(usage)
std = np.std(usage)

cv = (std / mean) * 100

print("Mean:", round(mean, 2))
print("Standard Deviation:", round(std, 2))
print("Coefficient of Variation:", round(cv, 2), "%")

if cv < 20:
    print("Interpretation: Data usage is consistent.")
else:
    print("Interpretation: Data usage is highly variable.")
```

#### 4. Suppose you are analyzing the number of likes on 10 recent Instagram posts: [120, 150, 130, 110, 170, 140, 135, 160, 125, 155]. Compute the standard deviation and compare it with the mean to comment on the consistency of engagement.

```python
import numpy as np

likes = np.array([120, 150, 130, 110, 170, 140, 135, 160, 125, 155])

mean = np.mean(likes)
std = np.std(likes)

print("Mean:", round(mean, 2))
print("Standard Deviation:", round(std, 2))

if std < mean:
    print("Interpretation: Engagement is fairly consistent.")
else:
    print("Interpretation: Engagement is highly variable.")
```

#### 5. Use ChatGPT or any AI tool to generate a small dataset (at least 8 values) representing daily UPI transaction amounts for a week. Then, calculate the variance and standard deviation of the generated data, and briefly interpret what the results mean for spending stability.

```python
import numpy as np

transactions = np.array([450, 520, 480, 510, 495, 530, 505, 490])

variance = np.var(transactions)
std_dev = np.std(transactions)

print("Transactions:", transactions)
print("Variance:", round(variance, 2))
print("Standard Deviation:", round(std_dev, 2))

print("\nInterpretation:")
print("A lower standard deviation indicates that daily spending")
print("is relatively stable with little variation.")
```
