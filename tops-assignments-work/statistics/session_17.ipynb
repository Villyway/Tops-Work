#### 2. Using any spreadsheet tool (Excel, Google Sheets) or Python (pandas), plot a scatter plot of Myntra ad spend (in lakhs) vs. number of app downloads for 10 sample data points. Visually draw or highlight the line of best fit on your plot.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data = pd.DataFrame({
    "Ad Spend (Lakhs)": [2,4,5,6,8,10,12,14,16,18],
    "Downloads": [20,28,35,40,52,60,70,82,90,100]
})

x = data["Ad Spend (Lakhs)"]
y = data["Downloads"]

plt.figure(figsize=(8,5))
plt.scatter(x, y, color="blue")

m, b = np.polyfit(x, y, 1)

plt.plot(x, m*x + b, color="red")

plt.title("Ad Spend vs Downloads")
plt.xlabel("Ad Spend (Lakhs)")
plt.ylabel("Downloads")
plt.grid(True)

plt.show()
```

#### 3. Using Python and scikit-learn, fit a simple linear regression model to predict the average daily orders on Swiggy based on daily advertising spend. Print out the slope (coefficient) and intercept, and explain in your own words what each means in this context.

**Hint:** The coefficient tells you how much orders increase for every extra lakh spent on ads.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = pd.DataFrame({
    "Ad Spend":[2,4,6,8,10,12,14,16],
    "Orders":[180,240,300,360,430,500,560,620]
})

X = data[["Ad Spend"]]
y = data["Orders"]

model = LinearRegression()

model.fit(X, y)

print("Slope (Coefficient):", round(model.coef_[0],2))
print("Intercept:", round(model.intercept_,2))

print("""
Explanation:
The coefficient represents the increase in daily orders
for every additional lakh spent on advertising.

The intercept represents the estimated number of orders
when advertising spend is zero.
""")
```

#### 4. Calculate the R-squared value for your regression model in the previous task and interpret what it tells you about the relationship between Swiggy's ad spend and daily orders.

**Hint:** Is the R-squared closer to 0 or 1? What does that mean for your model's accuracy?

```python
r2 = model.score(X, y)

print("R-Squared Value:", round(r2,4))

if r2 > 0.8:
    print("\nInterpretation:")
    print("The R-squared value is close to 1.")
    print("This indicates a strong relationship between ad spend and daily orders.")
else:
    print("\nInterpretation:")
    print("The R-squared value is closer to 0.")
    print("This indicates a weak relationship between ad spend and daily orders.")
```
